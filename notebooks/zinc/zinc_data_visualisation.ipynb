```python
#lag plot
from pandas.plotting import lag_plot
lag_plot(series)
plt.show()
```

```python
series = series.asfreq('MS')  # MS = Month Start
```

```python
#interpoler
series = series.interpolate(method='linear')
```

```python
print("Après lecture CSV:", len(series))
```

```python
plt.figure(figsize=(12,6))
plt.plot(series.index,series.values,marker='o',linewidth=2,color='steelblue')

plt.title('Évolution du prix du Zinc',fontsize=14)
plt.xlabel('Date')
plt.ylabel('Prix(USD/Tonne)')
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
