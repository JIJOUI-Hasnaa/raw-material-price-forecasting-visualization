```python
import pandas as pd
```

```python
data=pd.read_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\raw\zinc_history.csv")
```

```python
print(data.head())
```

```python
print(data.shape)
```

```python
print(data.info())
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
print(data.dtypes)
```

```python
print(data['Date'].head(10))
print(data['Date'].apply(repr).head(10))  # repr() montre les guillemets/espaces cachés
```

```python
for col in ['Price','Open','High','Low']:
    data[col]=data[col].astype(str).str.replace(',','',regex=False)
    data[col]=pd.to_numeric(data[col],errors='coerce')
data['Change %']=data['Change %'].astype(str).str.replace('%','',regex=False)
data['Change %']=pd.to_numeric(data['Change %'],errors='coerce')

#la colonne date en vrai datetime
data['Date']=pd.to_datetime(data['Date'],format='%m/%d/%Y',errors='coerce')
```

```python
print(data.dtypes)
```

```python
print((data.head()))
```

```python
data['Price']=data['Price'].fillna(data['Price'].mean())
```

```python
print(data['Date'].isna().sum())
```

```python
data=data.drop_duplicates(subset=['Date'])
```

```python
df = data[['Date', 'Price']].copy()
df.columns=('date','price')
print(df.head())
```

```python
df.to_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\processed\zinc_cleaned_data.csv",index=False)
```
