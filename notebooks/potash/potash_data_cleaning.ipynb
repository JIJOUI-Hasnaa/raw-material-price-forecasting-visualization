```python
months={
    "january": 1, "february": 2, "march": 3, "april": 4, 
    "may": 5, "june": 6, "july": 7, "august": 8, 
    "september": 9, "october": 10, "november": 11, "december": 12,

    "january.": 1, "february.": 2, "march.": 3, "april.": 4, 
    "may.": 5, "june.": 6, "july.": 7, "august.": 8, 
    "september.": 9, "october.": 10, "november.": 11, "december.": 12
}
```

```python
def parse_months(text):
    try:
        parts=str(text).strip().split()
        if len(parts)==3:
            mois,jour,annee=parts[0],parts[1],parts[2]
            num_mois=months.get(mois.lower())
            if num_mois:
                return pd.Timestamp(year=int(annee),month=num_mois,day=1)
    except Exception:
        pass
    return None


```

```python
print(parse_months("May 31, 2026"))
```

```python
df.loc[:,"Date"]=df["Date"].apply(parse_months)
print(df.head())
```

```python
df.dtypes
```

```python
# creer une copie indépendante au début pour bloquer les warnings
df = df.copy()

df["Date"] = pd.to_datetime(df["Date"], errors='coerce')
df["Value"] = pd.to_numeric(df["Value"], errors='coerce')

# verifiez que les types sont corrects
print(df.dtypes)
```
