```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\processed\sulfur_cleaned_data.csv",header=0)
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'], format='%Y')
df = df.set_index('date')
df.asfreq('YS')
df = df.squeeze()
print(df.head())
```

```python
df.describe()
```

```python
df.plot()
plt.show()
```

```python
df.plot(style='k.')
plt.show()
```

```python
df.hist()
plt.show()
```

```python
df.plot(kind='kde')
plt.show()
```

```python
from pandas.plotting import lag_plot
lag_plot(df)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df.index,df.values,marker='o',linewidth=2,color='purple')

plt.title("Évolution du prix du potassium",fontsize=14)
plt.xlabel('Année')
plt.ylabel('prix')
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
