```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import json
import time

# Activer les logs de performance pour capturer les requêtes réseau
options = Options()
options.set_capability("goog:loggingPrefs", {"performance": "ALL"})

driver = webdriver.Chrome(options=options)

try:
    driver.get("https://www.imarcgroup.com/sulphur-pricing-report")
    print("Chargement de la page et capture des requêtes...")
    time.sleep(8)  # Attendre que la requête XHR se déclenche
    
    # Récupérer les logs de performance du navigateur
    logs = driver.get_log("performance")
    
    for entry in logs:
        log_message = json.loads(entry["message"])["message"]
        
        # Filtrer pour trouver l'appel XHR/Fetch qui contient "getgraphbyreportid"
        if log_message["method"] == "Network.requestWillBeSent":
            url = log_message["params"]["request"]["url"]
            if "getgraphbyreportid" in url:
                print("\n[SUCCÈS] URL de l'API interne trouvée :")
                print(url)
                
                # Vous pouvez ensuite utiliser cette URL directement avec la bibliothèque `requests` 
                # (éventuellement en conservant les cookies de session du driver si nécessaire)

except Exception as e:
    print("Erreur :", e)

finally:
    driver.quit()
```

```python
import requests
import json

# L'URL exacte de l'API interne du graphique
url = "https://www.imarcgroup.com/mediaroom/getgraphbyreportid?Id=22674"

# Headers pour simuler un vrai navigateur et éviter d'être bloqué
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Referer": "https://www.imarcgroup.com/sulphur-pricing-report"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    try:
        data = response.json()
        print("Données récupérées avec succès !")
        # Afficher joliment le JSON
        print(json.dumps(data, indent=4, ensure_ascii=False))
        
        # Optionnel : Sauvegarder dans un fichier JSON
        with open("sulfur_prices_data.json", "w", encoding="utf-8") as f:
            json.dump(data, f, indent=4, ensure_ascii=False)
            print("\nDonnées sauvegardées dans 'sulfur_prices_data.json'")
            
    except json.JSONDecodeError:
        print("La réponse n'est pas au format JSON valide :")
        print(response.text)
else:
    print(f"Erreur HTTP : {response.status_code}")
```
