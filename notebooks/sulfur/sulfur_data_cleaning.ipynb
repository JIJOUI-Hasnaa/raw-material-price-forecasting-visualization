```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\raw\sulfur_annual_history.csv",skiprows=4,sep=';',header=0,parse_dates=True)
df=df.iloc[3:,:]
print(df.head())
```

```python
df.columns
```

```python
data=df[['Year','Unit value ($/t)']].copy()
print(data.head())
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
data.dtypes
```

```python
data['Year'] = data['Year'].astype(str).str.strip()
data['Year'] = pd.to_numeric(data['Year'], errors='coerce')
data.dtypes
```

```python
data = data.dropna(subset=['Year'])
data['Year']=data['Year'].astype(int)
data.dtypes
```

```python
#interpolation lineaire
data['Unit value ($/t)'] = data['Unit value ($/t)'].interpolate(method='linear')
```

```python
data.drop_duplicates(subset='Year')
```

```python
data.describe()
```

```python
data.columns=('date','price')
data.head()
```

```python
data.to_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\processed\sulfur_cleaned_data.csv",index=False)
```
