```python
df["price"]=df["Prix"].astype(str).str.replace(",",".").astype(float)
```

```python
frensh_months={
    "janv.": 1, "févr.": 2, "mars.": 3, "avr.": 4, 
    "mai.": 5, "juin.": 6, "juil.": 7, "août.": 8, 
    "sept.": 9, "oct.": 10, "nov.": 11, "déc.": 12,

    "janv": 1, "févr": 2, "mars": 3, "avr": 4, 
    "mai": 5, "juin": 6, "juil": 7, "août": 8, 
    "sept": 9, "oct": 10, "nov": 11, "déc": 12,
    

    "janv ": 1, "févr ": 2, "mars ": 3, "avr ": 4, 
    "mai ": 5, "juin ": 6, "juil ": 7, "août ": 8, 
    "sept ": 9, "oct ": 10, "nov ": 11, "déc ": 12,




}
```

```python
def parse_frensh_date(text):
    try:
        parts=str(text).strip().split()
        if len(parts)==2:
            mois,annee=parts[0],parts[1]
            num_mois=frensh_months.get(mois.lower())
            if num_mois:
                return pd.Timestamp(year=int(annee),month=num_mois,day=1)
    except Exception:
        pass
    return None
```

```python
df["Date"]=df["Mois"].apply(parse_frensh_date)
print(df.head())
```

```python
df.dtypes
```

```python
df["Date"]=pd.to_datetime(df["Date"],errors='coerce')
df['Prix']=pd.to_numeric(df['Prix'],errors='coerce')
df.dtypes
```
