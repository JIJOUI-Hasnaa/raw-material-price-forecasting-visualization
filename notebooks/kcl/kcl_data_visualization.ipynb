```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
series=pd.read_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\processed\kcl_cleaned_data.csv",header=0,index_col=0,parse_dates=True).squeeze()
series.describe()
```

```python
series=series.asfreq('MS')
```

```python
series.head(100)
```

```python
series.plot()
plt.show()
```

```python
series.plot(style='k.')
plt.show()
```

```python
series.hist()
plt.show()
```

```python
series.plot(kind='kde')
plt.show()
```

```python
from pandas.plotting import lag_plot
lag_plot(series)
plt.show()
```

```python
#interpolation
series=series.interpolate(method='linear')
```

```python
plt.figure(figsize=(12,6))
plt.plot(series.index,series.values,marker='o',linewidth=2,color='red')

plt.title("Évolution du prix de kcl ",fontsize=14)
plt.xlabel('Date')
plt.ylabel('Price')
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```
