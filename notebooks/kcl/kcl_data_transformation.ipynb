```python
import requests
import pandas as pd
```

```python
url="https://v6.exchangerate-api.com/v6/d0e80dadc76c6a0b8c8ca608/pair/INR/USD"
headers={
    "Authorization":"d0e80dadc76c6a0b8c8ca608",
    "Accept":"application/json"
}
#requete get
response=requests.get(url,headers=headers)
#status
if response.status_code!=200:
    print(f"Erreur :{response.status_code}")
    raise SystemExit(1)
   
data=response.json()
print(data)
```

```python
#transformation INR en USD
usd_rate=data['conversion_rate']
print(f"usd rate:{usd_rate}\n")
```

```python
df=pd.read_csv(r"C:\Users\Pc\Documents\pfa_stage_project\data\processed\kcl_cleaned_data.csv")
df.columns.tolist()
```

```python
df['Price'] = (df['Price'] * usd_rate).round(6)
df_out = df[['Date', 'Price']].rename(columns={'Price': 'usd_price'})
print(df_out.head())
# limite: ne reflète pas les fluctuations du taux INR/USD sur la période historique
```
